# **Question-3**

```python
from nltk.corpus import stopwords
from nltk.stem import LancasterStemmer, PorterStemmer, SnowballStemmer
from nltk.stem import WordNetLemmatizer
```

```python
movie_reviews.fileids("pos")
```

```python
review=movie_reviews.raw('pos/cv955_25001.txt')
print(review)
```

```python
words=word_tokenize(review)
lemmatizer=WordNetLemmatizer()
```

```python
print(words)
```

```python
stopWords=stopwords.words("english")
print(stopWords)
```

```python
flitered_review=[lemmatizer.lemmatize(w.lower()) for w in words if w.lower() not in stopWords]
print(flitered_review)
```

```python
wordCount= FreqDist(flitered_review)
mostFreq20=wordCount.most_common(20)
x=[i[0] for i in mostFreq20]
y=[i[1] for i in mostFreq20]
plt.figure(figsize=(12, 25))
plt.bar(x,y)
plt.title("Most Frequent 20 words")
plt.xlabel("Words")
plt.ylabel("Count")
```

# **Question-4**

```python
lincoln=inaugural.raw('1861-Lincoln.txt')
tokens=word_tokenize(lincoln)
```

```python
punctuations=string.punctuation
print(punctuations)
```

```python
filtered_speech=[i.lower() for i in tokens if i.lower() not in stopWords and i not in punctuations]
```

```python
print(filtered_speech)
```

```python
wordCount=FreqDist(filtered_speech)
wordCount15=wordCount.most_common(15)
print(wordCount15)
```

```python
top15=[i[0] for i in wordCount15]
print(top15)
```

```python
bigrams=list(ngrams(top15, n = 2))
bigrams
```

```python
unique_words=list(set(bigrams))
matrix=np.zeros((len(unique_words),len(unique_words)),dtype=int)
print(matrix)
```

```python
wordIndex={}
for i,word in enumerate(unique_words):
  wordIndex[word]=i
print(wordIndex)
```

```python
for i in range(len(filtered_speech)-1):
  current_word = filtered_speech[i]
  next_word = filtered_speech[i+1]
  if current_word in wordIndex and next_word in wordIndex:
        cw_index = wordIndex[current_word]
        nw_index = wordIndex[next_word]
        matrix[cw_index, nw_index] += 1

print("The Co-occurance matrix for the top 15 words:\n")
print(matrix)
```

# **Question-5**

```python
from nltk.corpus import udhr
```

```python
udhr.fileids()
```

```python
german=udhr.raw('German_Deutsch-Latin1')
print(german)
```

```python
words=word_tokenize(german)
print(words)
```

```python
vowels="aeiouAEIOUäöüÄÖÜ"
```

```python
filter=[]
for i in (words):
  vowel=""
  for y in i:
    if y in vowels and y!="":
      vowel+=y
  if vowel !="":
    filter.append(vowel)
print(filter)
```

```python
trigrams=list(ngrams(filter,n=3))
print(trigrams)
```

```python
freqCount=FreqDist(trigrams).most_common()
print(freqCount)
```
